```python
from pathlib import Path


def build_and_save_phone_averages(
    task,
    models,
    data_path,
    alignments,
    datasets,
    strategies,
    phone_classes,
    output_dir,
    utterance_mean_normalization=True,
    collapse_repeats=False,
):
    strategy_dir = {
        AveragingStrategy.CENTER_PHONE: "center_phone",
        AveragingStrategy.FULL_UNIT: "full_unit",
    }

    for encoder_name, info in models.items():
        for layer in info["layer"]:

            feature_root = (
                Path(data_path)
                / "paper_data"
                / "features"
                / encoder_name
                / task
                / f"layer{layer}"
            )

            for alignment_name, alignment_data in alignments.items():
                for dataset in datasets:

                    split_alignments = alignment_data[dataset]

                    for strategy in strategies:
                        for phone_class in phone_classes:

                            # FULL_UNIT + MONOPHONE is redundant.
                            if (
                                strategy == AveragingStrategy.FULL_UNIT
                                and phone_class == PhoneClass.MONOPHONE
                            ):
                                continue

                            strategy_name = strategy_dir[strategy]

                            save_path = (
                                Path(output_dir)
                                / encoder_name
                                / f"layer{layer}"
                                / alignment_name
                                / dataset
                                / strategy_name
                                / f"{phone_class.value}.pickle"
                            )
                            if save_path.exists():
                                # Most valid pickles should be substantially larger than this.
                                if save_path.stat().st_size > 1024:
                                    print(f"Skipping existing: {save_path}")
                                    continue
                            
                                # Only deserialize suspiciously small files.
                                existing = load_pickle(save_path)
                            
                                if len(existing) >0:
                                    print(f"Skipping existing: {save_path}")
                                    continue

                            print(
                                f"Processing {encoder_name} | "
                                f"layer={layer} | "
                                f"{alignment_name} | "
                                f"{dataset} | "
                                f"{strategy_name} | "
                                f"{phone_class.value}"
                            )

                            feature_path = feature_root / dataset

                            avf_feats = build_phone_averages_from_pickles(
                                feature_path=str(feature_path),
                                alignments=split_alignments,
                                phone_class=phone_class,
                                strategy=strategy,
                                utterance_mean_normalization=(
                                    utterance_mean_normalization
                                ),
                                collapse_repeats=collapse_repeats,
                            )


                            save_path.parent.mkdir(
                                parents=True,
                                exist_ok=True,
                            )

                            save_pickle(avf_feats, save_path)

                            print(f"Saved: {save_path}")
```

```python

build_and_save_phone_averages(
    task='sandi',
    models=models,
    common_path=DATA_PATH,
    alignments={
        #"ctc_wav2vec_timitv3": {
        #    "train": train_ctc_algn,
        #    "dev": dev_ctc_algn,
        #},
        "factored_hybrid_diphone": {
            "train": train_fh_algn_sandi,
            "dev": dev_fh_algn_sandi,
        },
    },
    datasets=["train", "dev"],
    strategies=[
        AveragingStrategy.CENTER_PHONE,
        AveragingStrategy.FULL_UNIT,
    ],
    phone_classes=[
        PhoneClass.MONOPHONE,
        PhoneClass.DIPHONE,
        PhoneClass.TRIPHONE,
    ],
    output_dir=f"{DATA_PATH}/phone_avg/sandi",
)
```
